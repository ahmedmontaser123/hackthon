```python
def calc_na_percentage(Data:pd.DataFrame) -> list:
    values = []
    for column in Data.columns:
        na = Data[column].isna().sum()/Data.shape[0]*100
        values.append(column)
        values.append(na)
    return values
```

```python
def Year_and_semester_extractor(Data: pd.DataFrame):
    df = Data.copy()
    df["presentation_year"] = df["code_presentation"].str[:4].astype(int)
    df["presentation_semester"] = df["code_presentation"].str[-1]
    df.drop(["code_presentation"],axis=1,inplace=True)
    return df
```

```python
df_AAA = Year_and_semester_extractor(df_AAA)
df_BBB = Year_and_semester_extractor(df_BBB)
df_CCC = Year_and_semester_extractor(df_CCC)
df_DDD = Year_and_semester_extractor(df_DDD)
df_EEE = Year_and_semester_extractor(df_EEE)
df_FFF = Year_and_semester_extractor(df_FFF)
df_GGG = Year_and_semester_extractor(df_GGG)
```

```python
from sklearn.preprocessing import OrdinalEncoder
def Ordinal_encoder_highest_education(Data : pd.DataFrame):
    df = Data.copy()
    edu_order = [
    "No Formal quals",
    "Lower Than A Level",
    "A Level or Equivalent",
    "HE Qualification",
    "Post Graduate Qualification"
    ]
    encoder = OrdinalEncoder(categories=[edu_order])
    df["highest_education_encoded"] = encoder.fit_transform(
    df[["highest_education"]]
    ) 
    return df.drop(['highest_education'],axis=1)
```

```python
df_AAA = Ordinal_encoder_highest_education(df_AAA)
df_BBB = Ordinal_encoder_highest_education(df_BBB)
df_CCC = Ordinal_encoder_highest_education(df_CCC)
df_DDD = Ordinal_encoder_highest_education(df_DDD)
df_EEE = Ordinal_encoder_highest_education(df_EEE)
df_FFF = Ordinal_encoder_highest_education(df_FFF)
df_GGG = Ordinal_encoder_highest_education(df_GGG)
```
